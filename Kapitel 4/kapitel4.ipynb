## *Kapitel 4 - Klassificering*

## **Faktafrågor**

### **Vad kännetecknar klassificeringsproblem? Ge några exempel på tillämpningsområden.**

Ett klassificeringsproblem innebär att vi vill använda en eller flera oberoende variabler, alltså features, för att bestämma vilken klass en observation tillhör. Den beroende variabeln y är alltså en klass eller ett label, i stället för ett kontinuerligt värde som i regression. Klassificering kan till exempel vara binär, där det finns två klasser, eller ha flera klasser. Ett exempel på binär klassificering är att förutsäga om en kund kommer att lämna ett företag eller stanna kvar. Andra exempel är att avgöra om ett mejl är spam eller inte spam, eller om en patient tillhör en viss riskklass. Klassificeringsmodeller kan också användas när vi har flera klasser, till exempel för att klassificera en bild av en handskriven siffra som 0, 1, 2 och så vidare. Det viktiga är alltså att modellens uppgift är att välja en klass för en observation utifrån den data som modellen har tränats på.

### **Förklara hur OvR- och OvO-algoritmerna fungerar.**

OvR betyder One-versus-Rest och används för att göra om ett problem med flera klasser till flera binära klassificeringsproblem. Om vi till exempel vill klassificera siffrorna 0 till 9 tränas en klassificerare för varje siffra. Den första modellen lär sig skilja 0 från alla andra siffror, den andra skiljer 1 från alla andra och så vidare. När en ny observation ska klassificeras körs modellerna och den klass som får högst score väljs som prediktion.

OvO betyder One-versus-One. Då tränas i stället en binär klassificerare för varje par av klasser. Om vi har siffrorna 0 till 9 får vi alltså modeller som jämför 0 mot 1, 0 mot 2, 0 mot 3 och så vidare. När en ny observation ska klassificeras får modellerna tävla mot varandra och den klass som vinner flest jämförelser blir prediktionen. Skillnaden är därför framför allt att OvR jämför varje klass mot alla andra klasser, medan OvO jämför klasserna två och två.

### **Förklara följande utvärderingsmått.**

En confusion matrix är ett sätt att sammanfatta hur en klassificeringsmodell har lyckats. För en binär klassificering kan vi få fyra olika resultat. 

True Positive, TP, betyder att modellen förutspådde den positiva klassen och hade rätt. 
True Negative, TN, betyder att modellen förutspådde den negativa klassen och hade rätt. 
False Positive, FP, betyder att modellen förutspådde den positiva klassen men hade fel. 
False Negative, FN, betyder att modellen förutspådde den negativa klassen men att observationen egentligen tillhörde den positiva klassen. 

Confusion matrix är extra användbar när klasserna är obalanserade, eftersom accuracy då kan ge en missvisande bild av hur bra modellen faktiskt är.
Accuracy beskriver hur stor andel av alla prediktioner som modellen klassificerade rätt. Det är alltså antalet korrekta prediktioner delat med det totala antalet prediktioner. Accuracy kan vara ett bra mått när klasserna är ganska jämnt fördelade, men kan bli missvisande när en klass är mycket vanligare än en annan.

Precision beskriver hur stor andel av de observationer som modellen förutspådde som positiva som faktiskt var positiva. Precision handlar alltså om hur mycket vi kan lita på modellens positiva prediktioner. Om modellen till exempel säger att 100 personer är positiva och 90 av dem verkligen är positiva är precisionen 90 procent. Precision blir särskilt viktig när falska positiva prediktioner är problematiska.

Recall beskriver i stället hur stor andel av alla verkligt positiva observationer som modellen lyckas hitta. Om det faktiskt finns 100 positiva observationer och modellen hittar 90 av dem är recall 90 procent. Recall blir särskilt viktigt när det är allvarligt att missa positiva fall.

F1-score kombinerar precision och recall till ett enda mått. F1-score är det harmoniska medelvärdet av precision och recall. För att få ett högt F1-score behöver modellen därför ha både hög precision och hög recall. Det är användbart när vi vill ta hänsyn till båda samtidigt.

ROC-kurvan visar hur modellens resultat förändras när vi ändrar gränsen för när en observation ska räknas som positiv. På y-axeln visas true positive rate, vilket motsvarar recall, och på x-axeln visas false positive rate. En modell som kan skilja klasserna bra får en ROC-kurva som ligger närmare det övre vänstra hörnet. ROC-kurvan hjälper alltså till att se hur bra modellen kan skilja mellan klasserna över olika beslutströsklar.

### **Vad är precision-recall tradeoff för något?**

Precision-recall tradeoff betyder att det ofta finns en avvägning mellan precision och recall. Om vi ändrar gränsen för när modellen ska kalla något positivt påverkar det hur många positiva fall modellen hittar och hur många av de positiva prediktionerna som faktiskt är rätt. Om vi gör modellen mer försiktig med att kalla något positivt kan precisionen öka, eftersom vi bara väljer de fall där modellen är mer säker. Samtidigt kan recall minska eftersom modellen då riskerar att missa fler verkligt positiva fall. Om vi i stället gör det lättare för modellen att kalla något positivt kan recall öka, men då kan precisionen sjunka eftersom fler falska positiva kommer med.

Det betyder att det inte finns ett enda läge som alltid är bäst. Vilken balans vi vill ha beror på vad modellen ska användas till och vilka konsekvenser olika typer av fel får. Därför behöver man välja en lämplig beslutströskel utifrån problemet.

### **Förklara översiktligt hur de vanligt förekommande klassificeringsmodellerna fungerar.**

Logistisk regression används för klassificering och kan bland annat ge en sannolikhet för att en observation tillhör en klass. Modellen använder de olika features som indata och beräknar en sannolikhet som sedan kan användas för att bestämma klass. Trots namnet är logistisk regression alltså en klassificeringsmodell. I scikit-learn används dessutom regularisering som standard, vilket hjälper till att begränsa modellen och minska risken för överanpassning.

Support Vector Machines, SVM, försöker hitta en gräns mellan klasserna. Tanken är att hitta en gräns som skiljer klasserna åt med så stor marginal som möjligt. SVM kan också använda så kallade kernels för att hantera situationer där klasserna inte kan skiljas åt med en enkel rak gräns. Modellen försöker då skapa en mer avancerad gräns mellan klasserna.

Ett beslutsträd fungerar genom att dela upp datan steg för steg utifrån olika features. Varje fråga eller villkor leder till en ny gren i trädet och till slut hamnar observationen i en lövnod där modellen bestämmer vilken klass den tillhör. Ett beslutsträd är därför relativt lätt att tolka eftersom man kan följa vilka beslut som ledde fram till prediktionen. Om trädet blir för komplext finns det däremot risk för overfitting.

Ensemble learning innebär att man kombinerar flera modeller i stället för att förlita sig på bara en modell. Tanken är att flera modeller tillsammans kan ge en bättre och mer stabil prediktion än en enskild modell. 
Ett exempel är VotingClassifier. Vid hard voting röstar modellerna om vilken klass som ska väljas. Vid soft voting används i stället modellernas sannolikheter för klasserna och dessa vägs samman.

Bagging är en annan typ av ensemble learning där flera modeller tränas på olika delmängder av träningsdatan. Vid bagging används sampling med återläggning, vilket gör att samma observation kan förekomma flera gånger i ett deldataset. Modellerna tränas sedan separat och deras prediktioner kombineras, till exempel genom majoritetsröstning vid klassificering. Tanken är att flera modeller tillsammans kan ge ett mer stabilt resultat än en enskild modell.

Random forest är en ensemblemodell som består av många beslutsträd. Träden tränas på olika slumpmässiga delar av datan och använder även slumpmässiga urval av features vid uppdelningarna. När modellen ska prediktera kombineras trädens resultat. På så sätt blir modellen mindre beroende av ett enda beslutsträd och kan bli mer robust. Random forest används därför ofta för att få bra prediktioner utan att behöva förlita sig på ett enda komplext träd.

Extra Trees, eller Extremely Randomized Trees, bygger också på många träd men använder ännu mer slumpmässighet när träden skapas. Bland annat väljs splitpunkterna mer slumpmässigt än i ett vanligt beslutsträd. Tanken är att den extra slumpmässigheten ska göra att träden blir mindre lika varandra och att kombinationen av dem kan ge en bra och stabil modell.

### **Vad innebär det att vi kan kolla på feature importance med hjälp av trädmodeller såsom beslutsträd eller random forest?**

Feature importance betyder att vi kan undersöka vilka features som modellen har haft störst nytta av när den gjort sina beslut. I ett beslutsträd kan vissa variabler användas flera gånger och bidra mycket till att dela upp observationerna i olika klasser. Dessa variabler kan då få högre importance. I en random forest kan man på liknande sätt sammanställa information från många träd för att få en bild av vilka features som varit viktigast för modellen.

Det kan vara användbart eftersom vi inte bara får veta vad modellen predikterar, utan också kan få en uppfattning om vilka variabler som verkar ha haft betydelse för prediktionerna. Det kan till exempel hjälpa oss att förstå datan, upptäcka vilka variabler som är mest användbara och eventuellt ta bort mindre relevanta variabler. Feature importance ska däremot inte automatiskt tolkas som att en feature orsakar resultatet. Den visar framför allt hur viktig variabeln har varit för just modellens beslut.

## **Resonemangfrågor**

### *Stina vill ha högsta möjliga precision. Vad händer då med recall? När kan hög precision vara bra eller dålig? Hur påverkas rättsväsendet?*

Om Stina vill ha så hög precision som möjligt behöver hon vara försiktig med att göra positiva prediktioner. Då kan hon minska antalet false positives, alltså fall där modellen säger att något är positivt fast det egentligen är negativt. Problemet är att modellen samtidigt kan missa fler verkligt positiva fall. Recall kan därför sjunka. Det är själva precision-recall tradeoff som Kalle frågar om.

Hög precision kan vara viktigt när en falsk positiv prediktion får stora konsekvenser. Ett exempel är ett system där vi skickar en person till en åtgärd bara om modellen bedömer personen som positiv. Om vi vill undvika att många oskyldiga eller felaktiga fall hamnar i den kategorin kan vi vilja ha hög precision. Samtidigt kan en väldigt hög precision vara dålig om det innebär att modellen missar många positiva fall. Då blir recall för låg.

Rättsväsendet är ett tydligt exempel där detta blir extra viktigt. Om en modell används som stöd för beslut som kan bidra till att en person döms och får fängelse kan false positives få mycket allvarliga konsekvenser. Vi vill då vara mycket försiktiga med att klassificera en oskyldig person som skyldig. Hög precision blir därför viktig, eftersom de positiva prediktionerna behöver vara mycket träffsäkra. Samtidigt kan vi inte bara maximera precision och ignorera recall. Om modellen blir för försiktig kan den missa personer som faktiskt är skyldiga. I ett sådant sammanhang måste man därför väga konsekvenserna av både false positives och false negatives mycket noggrant. En modell bör inte väljas enbart utifrån ett enda mått.

### **Förklara hur man kan tolka figur 4.8 på sidan 175.**

Figur 4.8 visar beslutsgränserna för en logistisk regressionsmodell som klassificerar observationer i två klasser, 0 och 1, utifrån variablerna x1 och x2. Bakgrundsfärgen visar modellens uppskattade sannolikhet för att en punkt tillhör klass 1. Färgskalan går från 0 till 1. De mörkare områdena motsvarar låg sannolikhet för klass 1, medan de gulare områdena motsvarar hög sannolikhet för klass 1.

Det går att se att sannolikheten förändras gradvis över figuren. När vi rör oss mot området där bakgrunden blir gul ökar modellens uppskattade sannolikhet för klass 1. När vi rör oss mot det mörklila området minskar sannolikheten. Den diagonala övergången mellan områdena kan ses som modellens beslutsgräns. Om vi använder en vanlig gräns på 0,5 kommer observationer på ena sidan av gränsen att klassificeras som klass 0 och observationer på den andra sidan som klass 1.

De olika punkterna visar de faktiska observationerna och deras klasser. Det är tydligt att klasserna inte är helt separerade. Det finns alltså punkter från olika klasser som ligger nära varandra och även på 'fel' sida av beslutsgränsen. Det visar att modellen inte kan klassificera alla observationer perfekt. Det är också en viktig del av figuren: logistisk regression hittar en gräns som beskriver mönstret i datan, men verklig data kan innehålla överlappning mellan klasserna.

Figuren visar därför både hur modellen gör sina klassificeringar och att modellen samtidigt kan ge sannolikheter. Det är inte bara ett hårt beslut mellan 0 och 1. Bakgrundsfärgen visar hur säker modellen är på att en punkt tillhör klass 1 i olika delar av området.

### **Varför använder vi .fit_transform() på träningsdatan men bara .transform() på validerings- och testdatan?**

Skillnaden handlar om att vi vill lära oss en preprocessing från träningsdatan och sedan använda exakt samma preprocessing på ny data. .fit_transform() gör två saker: först fit, där transformern lär sig det som behövs från träningsdatan, och sedan transform, där träningsdatan omvandlas med hjälp av det som lärts in. Det kan till exempel handla om att beräkna medelvärde och standardavvikelse vid standardisering.

På validerings och testdatan använder vi bara .transform(). Då använder vi de värden och regler som redan lärts från träningsdatan. Vi ska alltså inte köra fit igen på validerings- eller testdatan. Anledningen är att vi annars skulle låta information från den data som modellen ska utvärderas på påverka preprocessing-steget. Det skulle innebära en form av data leakage och göra utvärderingen mindre rättvis.

Tanken är därför att träningsdatan är den data som modellen får lära sig av. Valideringsdata används för att jämföra och justera modeller, och testdata används för den slutliga utvärderingen. För att utvärderingen ska vara rättvis ska preprocessing som kräver inlärning endast läras från träningsdatan och sedan appliceras på de andra datamängderna.
