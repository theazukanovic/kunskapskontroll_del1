# *Kapitel 3 - Regression*

## **Faktafrågor**

### **Vad kännetecknar regressionsproblem? Ge några exempel på tillämpningsområden**

Ett regressionsproblem handlar om att försöka förutsäga ett numeriskt värde, alltså ett tal. Den beroende variabeln y är därför ett numeriskt värde som kan anta många olika värden. Modellen försöker hitta ett samband mellan våra indata X och det värde vi vill förutsäga, så att modellen sedan kan göra en prediktion på ny data.

Det som skiljer regression från klassificering är framför allt vad vi vill få ut. Om vi till exempel vill förutsäga priset på ett hus är det regression, eftersom priset är ett numeriskt värde. Om vi däremot vill förutsäga om en person kommer att lämna ett företag eller stanna kvar, har vi ett klassificeringsproblem eftersom resultatet är olika klasser.

Exempel på regressionsproblem är att förutsäga huspriser utifrån storlek, läge och antal rum, förutsäga lön utifrån exempelvis antal år i yrket, förutsäga bränsleförbrukning, eller förutsäga priset på en bil eller diamant utifrån dess egenskaper. Det gemensamma är alltså att vi vill svara på frågan: ”Vilket numeriskt värde är mest rimligt att förutsäga?”

### **Förklara utvärderingsmåtten RMSE, MSE och MAE.**

RMSE, MSE och MAE används för att mäta hur stora fel modellen gör. Man jämför alltså modellens predikterade värden med de verkliga värdena. För alla tre gäller att lägre värde är bättre, eftersom ett lägre fel betyder att modellens prediktioner ligger närmare de riktiga värdena.

MSE, Mean Squared Error, innebär att man först räknar ut felet mellan det riktiga värdet och modellens prediktion, sedan kvadrerar felet och till sist tar medelvärdet. Eftersom felen kvadreras får stora fel mycket större påverkan. Om modellen till exempel gör några väldigt stora fel kommer MSE därför att öka ganska kraftigt.

RMSE, Root Mean Squared Error, är helt enkelt roten ur MSE. Den stora fördelen med RMSE är att resultatet hamnar i samma enhet som målvariabeln. Om vi till exempel försöker förutsäga huspriser i kronor kommer RMSE också att uttryckas i kronor. Det gör det lättare att förstå resultatet. RMSE kan ungefär tolkas som det genomsnittliga avståndet mellan modellens prediktioner och de riktiga värdena.

MAE, Mean Absolute Error, räknar i stället ut det genomsnittliga absoluta felet. Det betyder att vi inte bryr oss om ifall modellen ligger över eller under det riktiga värdet, utan bara hur långt ifrån den är. Ett fel på 10 blir alltså 10 oavsett riktning. MAE påverkas därför mindre av extrema fel än MSE och RMSE.

### **Om vi ska rangordna olika modeller, spelar det någon roll om RMSE eller MSE används? Varför?**

Nej, inte om vi jämför modeller på samma dataset. RMSE är roten ur MSE. Eftersom kvadratroten bevarar ordningen mellan positiva tal kommer den modell som har lägst MSE också att ha lägst RMSE. Därför blir själva rangordningen av modellerna densamma.

Det som däremot skiljer måtten åt är hur lätta de är att tolka. RMSE är oftast mer lättförståeligt eftersom det har samma enhet som målvariabeln. Om vi till exempel förutsäger huspriser och får ett RMSE på 50 000 kan vi tänka att modellens felstorlek ligger runt 50 000 kronor. MSE hamnar däremot i kvadrerade enheter, vilket gör det svårare att tolka direkt.


### **Förklara mycket översiktligt vad gradient descent är.**

Gradient descent är en optimeringsmetod som används för att hitta modellparametrar som gör modellens fel så litet som möjligt. Man kan tänka på det som att man står på ett berg där höjden representerar hur stort felet är. Målet är att ta sig nedåt tills man kommer till ett område där felet är så litet som möjligt.

I praktiken börjar modellen med vissa parametrar. Sedan undersöker gradient descent åt vilket håll felet minskar och justerar parametrarna lite åt det hållet. Detta upprepas steg för steg. Learning rate bestämmer ungefär hur stora stegen är. En för stor learning rate kan göra att man hoppar förbi minimum, medan en för liten kan göra processen väldigt långsam.


### **Vad är the bias variance trade-off? Varför är mer komplexa modeller inte alltid bättre?**

Bias och variance beskriver två olika problem som en modell kan ha. Hög bias betyder förenklat att modellen är för enkel och därför inte kan fånga viktiga mönster i datan. Modellen kan då underfitta. Hög variance betyder i stället att modellen är väldigt känslig för just den träningsdata som den fått och därför riskerar att även lära sig brus och oviktiga detaljer.

En enkel modell kan alltså ha hög bias och låg variance, medan en väldigt komplex modell kan ha låg bias men hög variance. En komplex modell kan passa träningsdatan väldigt bra, men problemet är att den kanske inte fungerar lika bra på ny data. Då har modellen overfittat.

Det är därför inte automatiskt bättre att göra modellen mer komplex. Målet är inte att få modellen att passa träningsdatan så perfekt som möjligt, utan att hitta en lagom komplex modell som generaliserar bra till data som modellen inte har sett tidigare. Det är själva trade-offen: vi försöker hitta en bra balans mellan bias och variance.

### **Förklara översiktligt hur respektive modell fungerar**

Linjär regression försöker beskriva sambandet mellan X och y med en linjär funktion, alltså ungefär som att hitta den linje som passar datan bäst. Modellen lär sig koefficienter för de olika features som tillsammans används för att göra en prediktion. I en enkel linjär regression kan modellen beskrivas som ŷ = β₀ + β₁x, där β₀ är interceptet och β₁ beskriver lutningen. Om x ökar med en enhet förändras den predikterade y ungefär enligt β₁.

Ridge är en form av linjär regression där man använder L2-regularisering. Det betyder att man lägger till ett extra straff i kostnadsfunktionen för stora koefficienter. Koefficienterna pressas därför mot noll. Tanken är att modellen inte ska bli onödigt komplex och att risken för overfitting ska minska.

Lasso använder i stället L1-regularisering. En viktig skillnad är att Lasso kan göra vissa koefficienter exakt noll. Det betyder att vissa variabler i praktiken tas bort från modellen. Därför kan Lasso även användas för variabelselektion.

Elastic Net är en kombination av Ridge och Lasso. Den använder alltså både L1 och L2 regularisering. På så sätt kan modellen både krympa koefficienter och sätta vissa av dem till noll.

SVR, Support Vector Regression, är SVM för regression. Modellen försöker hitta en funktion där så många observationer som möjligt hamnar inom en toleranszon runt funktionen. Man kan tänka på det som ett rör runt modellens linje. ε bestämmer hur bred toleranszonen är och C styr hur hårt fel utanför zonen straffas. SVR kan också använda kernels för att fånga mer komplexa, icke-linjära samband.

Ett beslutsträd delar upp datan steg för steg genom olika frågor eller villkor. Exempelvis kan modellen först fråga om miltalet är över 8 000 och sedan ställa ytterligare frågor om ålder eller motor. Till slut hamnar observationen i ett löv, där modellen ger sin prediktion. Fördelen är att träd kan fånga icke-linjära samband, men ett väldigt djupt träd kan lätt överanpassa träningsdatan.

Ensemble learning innebär att man kombinerar flera modeller i stället för att bara använda en. Tanken är att flera modeller tillsammans kan ge en mer stabil och ofta bättre prediktion än en enskild modell.

Random Forest är en ensemble som består av många beslutsträd. Träden tränas på olika bootstrap-samplade delar av datan och använder slumpmässiga urval av features. Vid regression kombineras trädens prediktioner vanligtvis genom att ta medelvärdet. Det gör att Random Forest ofta blir mer robust än ett enskilt beslutsträd.


### **Vad menas med white box modeller och black box modeller?**

En white box-modell är en modell som är relativt lätt att förstå och tolka. Man kan i större utsträckning se hur modellen kommer fram till sitt resultat. Linjär regression är ett bra exempel eftersom koefficienterna kan ge information om hur olika features påverkar prediktionen.

En black box-modell är däremot svårare att förstå på detaljnivå. Modellen kan vara väldigt bra på att göra prediktioner, men det kan vara svårt att förklara exakt varför modellen kom fram till ett visst resultat. Mer komplexa ensemblemodeller och neurala nätverk kan till exempel vara svårare att tolka.

Det betyder inte att white box alltid är bättre. Om vi exempelvis arbetar inom ett område där det är väldigt viktigt att kunna förklara varför modellen fattade ett visst beslut kan en white box-modell vara extra värdefull. Om det viktigaste i stället är att få så bra prediktioner som möjligt kan en mer komplex modell ibland vara bättre.

### **Vad är skillnaden mellan bagging och pasting?**

Både bagging och pasting innebär att man tränar flera modeller på olika delmängder av träningsdatan och sedan kombinerar modellernas resultat. Skillnaden är framför allt hur observationerna väljs.

Vid bagging används sampling med återläggning. Det betyder att samma observation kan väljas flera gånger och därför förekomma flera gånger i samma deldataset.

Vid pasting används sampling utan återläggning. En observation kan därför bara förekomma en gång i varje deldataset.

Bagging kan på så sätt skapa flera lite olika träningsdataset. När modellerna sedan kombineras kan resultatet bli mer stabilt och variansen kan minska. Random Forest bygger bland annat på bagging-idén.


## **Resonemangfrågor**

 ### **Förklara hur man kan tolka figur 3.1 på sidan 113.**

Figur 3.1 visar en enkel linjär regressionsmodell där vi försöker förutsäga en persons inkomst utifrån personens ålder. På x-axeln finns ålder (x) och på y-axeln finns inkomst (y). De svarta punkterna är de faktiska observationerna, alltså de verkliga värdena på inkomsten för olika personer.

Den blå linjen är den linjära regressionsmodellens förutsägelse. Den kan skrivas som

$$
\hat{y} = \hat{\theta}_0 + \hat{\theta}_1x
$$

där $\hat{\theta}_0$ är modellens intercept och $\hat{\theta}_1$ beskriver hur mycket den förväntade inkomsten förändras när åldern ökar med en enhet. I figuren ser vi att linjen lutar uppåt, vilket betyder att modellen förväntar sig att inkomsten ökar när åldern ökar.

För varje observation kan vi jämföra det verkliga värdet $y_i$ med modellens förutsagda värde $\hat{y}_i$. Skillnaden mellan dessa värden kallas residual, eller fel, och skrivs:

$$
e_i = y_i - \hat{y}_i
$$

Det är de röda vertikala linjerna i figuren. Om en punkt ligger över regressionslinjen är residualen positiv, eftersom det verkliga värdet är högre än modellens förutsägelse. Om punkten ligger under linjen blir residualen negativ.

Figuren visar alltså väldigt tydligt vad regression handlar om, modellen försöker hitta en linje som beskriver sambandet mellan ålder och inkomst så bra som möjligt. Modellen kommer inte att träffa alla svarta punkter exakt, utan det kommer nästan alltid att finnas ett fel mellan det verkliga värdet och det förutsagda värdet. Modellen försöker hitta den linje som ger så små fel som möjligt för observationerna, samtidigt som modellen ska kunna generalisera till nya observationer.

Kort sagt: de svarta punkterna är de verkliga värdena, den blå linjen är modellens förutsägelser och de röda linjerna visar residualerna, alltså skillnaden mellan verkligt och förutsagt värde.


### **Förklara hur man kan tolka figur 3.13 på sidan 140. Hur hänger den ihop med figur 3.14 på sidan 141?**

Figur 3.13 visar hur ett beslutsträd för regression kan se ut. Till skillnad från den linjära regressionen i figur 3.1 försöker beslutsträdet inte beskriva sambandet med en enda rak linje. Istället delar trädet upp datan i olika grupper genom att ställa frågor om variablerna.

I det här exemplet börjar trädet med frågan:

$$
x_2 \leq 0.438?
$$

Om svaret är ja går observationen åt vänster och om svaret är nej går den åt höger. Därefter kan trädet ställa ytterligare frågor, till exempel $x_2 \leq -0.755$ eller $x_1 \leq -0.654$. Till slut hamnar observationen i ett blad, där modellen ger ett förutsagt värde.

I varje nod visas bland annat samples, squared_error och value.

Samples visar hur många observationer som finns i den aktuella noden. Value är det värde som modellen förutsäger för observationerna i den noden. Squared_error visar hur stora de kvadrerade felen är i noden. Ett lägre värde innebär att observationernas y-värden ligger mer samlat kring det värde som modellen förutsäger. När trädet delar upp datan försöker det göra grupperna så homogena som möjligt, alltså grupper där de observerade y-värdena ligger närmare varandra.

Figur 3.14 visar sedan samma beslutsträd på ett annat sätt. Istället för att visa själva grenarna och frågorna visuellt visar figuren vilka områden i $x_1$ och $x_2$ planet som hör till de olika löven i trädet. De horisontella och vertikala gränserna i figuren kommer alltså från frågorna som beslutsträdet ställer.

Till exempel finns i figur 3.13 frågan:

$$
x_2 \leq 0.438
$$

Det skapar en gräns vid $x_2 = 0.438$ i figur 3.14. På samma sätt skapar frågan

$$
x_2 \leq -0.755
$$

en annan gräns. På den högra sidan av trädet finns istället frågan

$$
x_1 \leq -0.654
$$

vilket skapar en vertikal gräns vid $x_1 = -0.654$.

Det är därför figur 3.13 och 3.14 hör ihop. Figur 3.13 visar själva beslutsträdet och vilka beslut som tas, medan figur 3.14 visar resultatet av dessa beslut som olika områden i diagrammet.

Man kan till exempel se att figur 3.13 har fyra slutnoder/blad. Därför delas området i figur 3.14 upp i fyra olika områden. Alla observationer som hamnar i samma område får samma förutsagda värde från beslutsträdet.

Det här visar också en viktig skillnad mot linjär regression. En linjär regression ger en rak linje och kan därför vara ganska begränsad när sambandet är mer komplicerat. Ett beslutsträd kan istället skapa olika områden och fånga mer komplexa samband mellan variablerna.

Samtidigt finns det en risk med beslutsträd. Om trädet får bli väldigt djupt kan det dela upp datan väldigt mycket och börja lära sig detaljer och brus i träningsdatan. Då kan modellen få väldigt bra resultat på träningsdatan men sämre resultat på nya data. Det kallas överanpassning (overfitting). Därför behöver man begränsa trädets komplexitet, exempelvis genom `max_depth`.

### **Förklara vad determinationskoefficienten R² är**

R², eller determinationskoefficienten, är ett mått som beskriver hur mycket av variationen i den beroende variabeln y som modellen kan förklara jämfört med en enkel baslinjemodell som alltid använder medelvärdet av y som prediktion.

Ett R² på 1 betyder att modellen förklarar variationen perfekt. Ett R² på 0 betyder ungefär att modellen inte är bättre än baslinjen där vi alltid skulle ha gissat medelvärdet. R² kan även bli negativt om modellen fungerar sämre än denna enkla baslinje.

Det är viktigt att inte blanda ihop R² med accuracy. Om en modell har R² = 0,80 betyder det inte att modellen har 80 procent rätt. Det betyder ungefär att modellen förklarar 80 procent av variationen i målvariabeln relativt baslinjen.

R² kompletterar därför mått som RMSE och MAE. RMSE och MAE berättar hur stora felen är, medan R² ger information om hur mycket variation i y modellen kan förklara.
