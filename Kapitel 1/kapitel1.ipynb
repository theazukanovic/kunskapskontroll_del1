## **Faktafrågor**

 ### **Hur hänger AI, ML och DL ihop?**

AI, alltså artificiell intelligens, är det bredaste begreppet. Det handlar om att skapa system som kan utföra uppgifter som vi normalt förknippar med mänsklig intelligens. Det kan till exempel vara att känna igen bilder, förstå språk, hitta mönster, fatta beslut eller lösa problem.

Machine Learning (ML) är en del av AI. Skillnaden är att man inom Machine Learning låter modellen lära sig mönster från data i stället för att vi behöver programmera alla regler själva. Man matar alltså modellen med exempel och tränar den på dessa. När modellen sedan möter ny data kan den använda det den har lärt sig för att göra en förutsägelse eller klassificering.

Deep Learning (DL) är i sin tur en del av Machine Learning. Deep Learning bygger på neurala nätverk med flera lager och kan lära sig mer komplexa mönster. Det används till exempel mycket inom bild-, text- och talanalys.

Man kan därför tänka på det som en hierarki: AI är det stora området, ML är en del av AI och DL är en del av ML. Alla Deep Learning-modeller är alltså Machine Learning, och Machine Learning är en del av AI, men all AI behöver inte använda Machine Learning eller Deep Learning.

Ett enkelt exempel är bildigenkänning. Ett AI-system kan ha som uppgift att känna igen vad som finns på en bild. Om systemet använder Machine Learning lär det sig detta från många exempelbilder. Om man använder Deep Learning kan ett neuralt nätverk själv lära sig många av de mönster som behövs för att skilja exempelvis hundar från katter.

### **Vilka är de fyra problemkategorierna inom ML?**

De fyra problemkategorierna som tas upp är regression, klassificering, dimensionsreduktion och klustring. Skillnaden mellan dem handlar framför allt om vilken typ av problem vi försöker lösa och vilket resultat vi vill få ut från modellen.

Regression används när vi vill förutsäga ett numeriskt värde. Ett exempel är att förutsäga priset på en bil utifrån exempelvis årsmodell, miltal och motortyp. Resultatet kan då vara exempelvis 145 000 kronor.

Klassificering används när modellen ska avgöra vilken kategori något tillhör. Ett exempel är att avgöra om ett mejl är spam eller inte spam. Till skillnad från regression är resultatet här en kategori i stället för ett kontinuerligt numeriskt värde.

Dimensionsreduktion innebär att man minskar antalet variabler eller dimensioner i datan samtidigt som man försöker behålla den viktigaste informationen. Det kan vara användbart om datan innehåller väldigt många variabler och därför är svår att analysera eller visualisera.

Klustring används för att hitta grupper i data. Här har vi inte nödvändigtvis bestämt grupperna i förväg, utan modellen försöker hitta observationer som liknar varandra. Ett exempel kan vara att dela in kunder i grupper baserat på deras köpbeteende.

Det viktiga är alltså att först förstå vad vi vill att modellen ska göra. Vill vi förutsäga ett tal passar regression, vill vi välja en kategori passar klassificering, vill vi förenkla många dimensioner kan dimensionsreduktion vara relevant och vill vi hitta naturliga grupper kan klustring användas.

### **Förklara följande:**

### **3a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?**

När vi bygger en ML-modell vill vi inte bara veta hur bra den fungerar på den data den har tränats på. Vi vill framför allt veta hur bra den fungerar på ny data som den inte har sett tidigare. Därför delar man upp datan i olika delar med olika roller.

Träningsdata används för att träna modellen. Det är här modellen får lära sig sambanden och mönstren i datan. Modellens parametrar anpassas utifrån träningsdatan.

Valideringsdata används under utvecklingen av modellen. Den kan till exempel användas för att jämföra olika modeller eller olika inställningar av hyperparametrar. På så sätt kan vi fatta beslut om modellen utan att använda testdatan.

Testdata sparas till slutet. Den används för att göra en slutlig utvärdering på data som modellen inte har använt när vi valt modell eller inställningar. Det ger en mer rättvis bild av hur modellen kan fungera på verkliga, nya observationer.

En viktig anledning till detta är att en modell kan bli överanpassad. Den kan vara väldigt bra på träningsdatan eftersom den har lärt sig just den datan, men fungera sämre på nya observationer. Genom att hålla isär träning, validering och test kan vi lättare upptäcka detta och få en bättre uppfattning om modellens förmåga att generalisera.

### **b) Vad är k-delad korsvalidering för något?**

K-delad korsvalidering, eller k-fold cross-validation, är en metod för att utvärdera en ML-modell. I stället för att bara göra en enda uppdelning av träningsdatan delar man upp den i k ungefär lika stora delar, som kallas folds.

Om vi till exempel använder 5-fold cross-validation delas datan upp i fem delar. Modellen tränas på fyra delar och valideras på den femte. Sedan byter man vilken del som används för validering. Detta fortsätter tills alla fem delar har fått vara valideringsdata en gång.

Man får alltså flera resultat i stället för bara ett. Resultaten kan sedan sammanfattas, exempelvis genom att ta ett medelvärde. Det ger en mer stabil bild av modellens prestation än om resultatet bara beror på en enda slumpmässig uppdelning.

Metoden är särskilt användbar när man har begränsat med data, eftersom varje observation kan vara med och bidra till både träning och validering under olika omgångar. Samtidigt ska testdatan fortfarande hållas separat om man använder en separat testuppsättning.

Anledningen till att vi använder cross-validation är alltså att vi vill få en mer pålitlig utvärdering och minska risken att vi råkar välja en modell bara för att den fungerade bra på en viss uppdelning av datan.


### **c) Vad är root mean squared error (RMSE) för något?**

RMSE står för Root Mean Squared Error och är ett mått som används för att utvärdera hur bra en regressionsmodell gör sina prediktioner. Det mäter hur stora felen mellan modellens prediktioner och de faktiska värdena är.
Först räknar man ut felen, sedan kvadreras felen så att negativa och positiva fel inte tar ut varandra. Därefter tar man medelvärdet och till sist roten ur det värdet. Därför kallas det Root Mean Squared Error.
RMSE är i samma enhet och skala som det vi försöker förutsäga, vilket gör det lättare att tolka. Ett lägre RMSE betyder att modellen gör mindre fel och därför är bättre utifrån detta mått.


### **d) Vad är en hyperparameter för något? Vad är en parameter för något?**

En parameter är ett värde som modellen själv lär sig från träningsdatan. Parametrarna bestäms alltså under träningen när modellen försöker hitta ett samband mellan input och det vi vill förutsäga. I en linjär regression är exempelvis modellens koefficienter parametrar.

En hyperparameter är i stället en inställning som vi bestämmer innan eller under modellutvecklingen. Modellen lär sig inte själv vilket hyperparametervärde vi ska använda på samma sätt som den lär sig sina parametrar. Exempel på hyperparametrar kan vara max_depth i ett beslutsträd eller antalet grannar i KNN.

Skillnaden kan därför förenklas till att modellen lär sig parametrarna från datan, medan vi väljer hyperparametrarna. Vi kan också använda metoder som Grid Search för att testa flera olika hyperparametrar och hitta en kombination som fungerar bra.

Det är viktigt att skilja på dem eftersom vi annars kan blanda ihop vad modellen faktiskt lär sig och vad vi själva ställer in. Hyperparametrarna kan dessutom påverka hur komplex modellen blir och därmed påverka risken för exempelvis overfitting.


### **e) Vad är grid search? Vad innebär refit=True?**

Grid search är en metod för att hitta bra värden på en modells hyperparametrar. Man anger olika möjliga värden för hyperparametrarna, vilket bildar ett slags "grid" (rutnät) av olika kombinationer. Modellen tränas och utvärderas för kombinationerna för att hitta den som ger bäst resultat.

När refit=True används tränas modellen om med de bästa hittade hyperparametrarna på hela träningsdatan som skickades in till GridSearchCV. Den färdigtränade modellen kan sedan användas för att göra prediktioner.


### **f) Vad är kategorisk data och hur hanteras det?**

Kategorisk data är data som består av olika kategorier istället för numeriska värden. Nominal data är kategorier som inte har någon särskild ordning, exempelvis olika färger. Ordinal data har däremot en ordning, exempelvis låg, medel och hög.
Eftersom många maskininlärningsmodeller behöver numerisk data kan kategorierna behöva omvandlas. Med one-hot encoding får varje kategori en egen kolumn med värdet 0 eller 1. Om vi till exempel har tre kategorier får vi tre nya kolumner.

Dummy-variable encoding fungerar på liknande sätt, men en av kategorierna tas bort. Om vi har k kategorier får vi därför k−1 kolumner. 

Ordinal encoding används när kategorierna har en naturlig ordning. Då ersätts kategorierna med numeriska värden utifrån ordningen, exempelvis låg = 1, medel = 2 och hög = 3.
Det är viktigt att välja encoding utifrån vilken typ av kategorisk data man har.


### **g) Vad är feature engineering för något?**

Feature engineering innebär att man skapar, omformar eller väljer features, alltså variabler som modellen använder, för att göra datan mer användbar för ML-modellen.

Tanken är att vi försöker ge modellen bättre information om problemet. Modellen kan bara lära sig från den data den får, så om en viktig egenskap saknas kan det vara svårt för modellen att hitta ett bra samband.

Ett enkelt exempel är bilpriser. Om vi har bilens årsmodell kan vi skapa en ny feature för bilens ålder. Bilens ålder kan i vissa fall vara mer direkt användbar för modellen än själva årsmodellen.

Feature engineering kan också innebära att kombinera eller omvandla befintliga variabler. Det kan exempelvis vara att skapa en kvot, gruppera värden eller omvandla en variabel till ett format som modellen lättare kan använda.

Det är därför en viktig del av ett ML-projekt. Det handlar inte bara om att välja en avancerad algoritm. Om vi ger modellen bättre och mer relevant information kan en enklare modell ibland prestera bättre än en avancerad modell som får dåliga features.


### **h) Vad menas med principle of parsimony?**

Principle of parsimony innebär att man om två modeller presterar ungefär lika bra, bör välja den enklare modellen. Det handlar alltså om att inte göra modellen mer komplex än vad som behövs för att lösa problemet.
En enklare modell kan till exempel ha färre variabler eller färre komplexa delar. Tanken är att en onödigt komplex modell kan bli svårare att tolka och kan öka risken för overfitting.
Principen kan sammanfattas som att man ska välja den enklaste modellen som ger ett tillräckligt bra resultat.

### **Vad menas med att “en modell är en förenkling av verkligheten”?**

En ML-modell är en förenkling av verkligheten eftersom verkligheten är mycket mer komplex än den data och de variabler som modellen får tillgång till. Modellen försöker hitta de viktigaste mönstren och sambanden för den uppgift vi vill lösa.

Om vi exempelvis bygger en modell som ska förutsäga priset på en bil kan vi använda information som årsmodell, miltal, motor och bilmärke. I verkligheten påverkas priset även av sådant som bilens exakta skick, efterfrågan på marknaden, hur säljaren presenterar bilen och många andra faktorer som kanske inte finns i datan.

Modellen behöver därför inte beskriva verkligheten perfekt för att vara användbar. Den ska skapa en tillräckligt bra bild av de samband som är relevanta för problemet.

Det här är också en viktig anledning till att modellens resultat aldrig bör ses som en absolut sanning. En modell gör en uppskattning baserad på den information och de mönster som den har fått lära sig från. Om verkligheten förändras eller om viktig information saknas kan modellens förutsägelser bli sämre.

### **Vad menas med att en modell är överanpassad/overfitted?**

Overfitting betyder att modellen har anpassat sig för mycket till träningsdatan. Den lär sig inte bara de verkliga mönstren utan även sådant som kan vara brus, slump eller väldigt specifikt för just träningsdatan.

Ett vanligt tecken på overfitting är att modellen presterar väldigt bra på träningsdata men betydligt sämre på validerings- eller testdata. Modellen verkar alltså vara bra när den får se data som liknar det den tränat på, men har svårt att generalisera till nya observationer.

Man kan tänka på det som att modellen nästan har 'pluggat in' träningsdatan i stället för att lära sig de generella reglerna bakom den. Om vi sedan ger modellen nya exempel som den aldrig sett tidigare fungerar den därför sämre.

Overfitting kan bland annat motverkas genom att minska modellens komplexitet, använda cross-validation, välja bättre hyperparametrar eller använda mer träningsdata. Målet är att hitta en modell som lär sig tillräckligt mycket för att fånga de riktiga sambanden, men inte så mycket att den börjar lära sig brus.

Det viktiga är alltså inte att få modellen att prestera maximalt på träningsdatan. Det viktiga är att den ska kunna generalisera och fungera bra även på ny data.

### **Högre är bättre i scikit-learn scoring, vad innebär det?**

Scikit-learn har en generell princip för scoring där ett högre score ska betyda att modellen är bättre. Det gör att olika scoring-metoder kan användas på ett enhetligt sätt när modeller ska jämföras eller hyperparametrar ska väljas.

Problemet är att vissa mått egentligen fungerar tvärtom. För exempelvis Mean Squared Error (MSE) är ett lägre värde bättre eftersom det betyder att modellens fel är mindre.

För att passa in i scikit-learns scoring-system kan ett sådant felmått därför göras negativt. Då blir exempelvis -10 bättre än -50, eftersom -10 motsvarar ett mindre MSE.

Det är därför viktigt att förstå vad score-värdet faktiskt representerar. Ett negativt score betyder inte automatiskt att modellen är dålig. Man behöver veta vilket scoring-mått som används och hur scikit-learn har definierat det.

Den här principen blir extra viktig när man använder exempelvis GridSearchCV. Grid Search behöver veta vilken kombination som är bäst, och genom att använda en scoring-konvention där högre alltid är bättre kan den välja det högsta score-värdet på ett konsekvent sätt.

### **Vad är tvärsnittsdata, tidsseriedata och paneldata?**

Tvärsnittsdata, eller cross-sectional data, innebär att man har information om flera olika observationer vid ungefär samma tidpunkt. Det kan exempelvis vara information om 1 000 personer under ett visst år, där varje person är en observation.

Tidsseriedata, eller time-series data, innebär att man följer en variabel eller en enhet över tid. Ett exempel är ett företags försäljning månad för månad under fem år. Här är tidsordningen viktig eftersom tidigare värden kan ha betydelse för senare värden.

Paneldata kombinerar de två perspektiven. Då följer man flera enheter över flera tidpunkter. Ett exempel kan vara att man följer 100 företag och deras omsättning varje år under tio år.

Skillnaden kan därför sammanfattas så här: tvärsnittsdata fokuserar på flera enheter vid en tidpunkt, tidsseriedata fokuserar på förändring över tid och paneldata följer flera enheter över tid.

### **Ge några exempel på verkliga tillämpningsområden inom ML.**

Machine Learning används i många verkliga situationer där det finns mycket data och där man vill hitta mönster, göra förutsägelser eller automatisera beslut. Några exempel är sjukvård, finans, e-handel, transport samt bild- och textanalys.

Inom sjukvården kan Machine Learning exempelvis användas för att analysera medicinska bilder eller hitta mönster i patientdata. En modell kan tränas på tidigare exempel och sedan hjälpa till att identifiera mönster i nya fall. Det betyder inte att modellen automatiskt ersätter en läkare, utan den kan fungera som ett stöd.

Inom finans kan ML användas för exempelvis riskbedömning och för att upptäcka misstänkta eller ovanliga transaktioner. Modellen kan lära sig vilka mönster som brukar förekomma och sedan reagera när en ny transaktion avviker från dessa mönster.

Inom e-handel används ML bland annat i rekommendationssystem. Om en kund har tittat på eller köpt vissa produkter kan modellen använda tidigare beteenden för att föreslå andra produkter som kunden kan vara intresserad av.

Inom transport kan ML användas för att analysera trafikmönster och göra prognoser. Historiska data kan till exempel hjälpa en modell att förutsäga trafik eller efterfrågan vid olika tidpunkter.

Bild och textanalys är ytterligare stora användningsområden. Modeller kan exempelvis klassificera bilder eller analysera text. Deep Learning används ofta när datan är komplex, till exempel vid bild och språkproblem.

Gemensamt för exemplen är att modellen får lära sig från tidigare data och sedan använder det den har lärt sig på nya situationer. Vilken ML-metod som passar beror på problemet. Om vi vill förutsäga ett tal kan regression passa, om vi vill välja en kategori kan klassificering passa och om vi vill hitta grupper utan färdiga kategorier kan klustring vara relevant.

### **Varför används scoring='neg_mean_squared_error'?**

Mean Squared Error (MSE) är ett felmått som mäter hur långt modellens förutsägelser ligger från de faktiska värdena. Eftersom vi vill ha så små fel som möjligt är ett lägre MSE bättre.

Scikit-learn använder däremot principen att ett högre scoring-värde alltid ska vara bättre. För att kunna använda MSE inom den principen använder scikit-learn därför en negativ version: neg_mean_squared_error.

Det betyder i praktiken att MSE får ett minustecken framför sig. Om en modell har MSE = 10 blir scoring-värdet -10. Om en annan modell har MSE = 50 blir scoring-värdet -50.

Eftersom -10 är högre än -50 kommer scikit-learn att betrakta modellen med MSE = 10 som bättre. Det är logiskt eftersom den modellen faktiskt har det mindre felet.

Det är därför man använder 'negative' i namnet. Det betyder inte att vi egentligen vill ha ett negativt fel. Det är bara ett sätt att vända riktningen så att MSE passar scikit-learns regel att högre scoring är bättre.

När man läser resultatet måste man därför komma ihåg detta. Om vi använder neg_mean_squared_error är ett score närmare noll bättre. För att tolka själva MSE-värdet som ett vanligt felmått behöver man tänka bort minustecknet.
