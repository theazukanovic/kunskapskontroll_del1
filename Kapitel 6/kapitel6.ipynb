# *Kapitel 6 - Klustring*

## **Faktafrågor**

### **Vad är klustring för något? Ge några exempel på tillämpningsområden.**

Klustring är en metod inom maskininlärning där man delar in observationer i olika grupper, alltså kluster. Till skillnad från klassificering har man normalt inga färdiga etiketter som talar om vilken grupp en observation tillhör. Modellen försöker i stället hitta grupper själv genom att leta efter observationer som liknar varandra. Tanken är alltså att observationer inom samma kluster ska vara mer lika varandra än observationer som ligger i andra kluster.

Klustring kan användas inom många olika områden. Ett exempel är kundsegmentering, där ett företag kan dela in sina kunder efter till exempel ålder och inkomst för att se vilka grupper som finns. Företaget kan sedan anpassa sin marknadsföring och sina erbjudanden efter de olika grupperna. Klustring kan också användas för att hitta avvikande observationer, till exempel misstänkta eller ovanliga transaktioner som ligger långt från de övriga observationerna. Ett annat exempel är semi-väglett lärande, där man först använder klustring för att hitta grupper i data och sedan kan använda grupperna för att få fram information som kan hjälpa en annan ML-modell.

### **Hur fungerar K-means? Använd figur 6.3 och figur 6.4 i din förklaring.**

K-means är en klustringsalgoritm som försöker dela upp data i ett bestämt antal kluster. Man bestämmer först hur många kluster man vill ha, alltså värdet på k. Därefter börjar algoritmen med att placera ut k stycken centroidpunkter. En centroid kan ses som centrumet för ett kluster.

Sedan räknar algoritmen ut vilket centroid varje observation ligger närmast och tilldelar observationerna till det kluster som har närmast centroid. När observationerna har delats upp räknas centroidernas positioner om utifrån observationerna i respektive kluster. Därefter görs samma sak igen: observationerna tilldelas det närmaste centroidet och centroidernas positioner uppdateras. Detta fortsätter tills modellen har konvergerat, alltså tills förändringarna blir tillräckligt små.

Figur 6.3 visar att K-means kan fungera bra när klustren faktiskt har en form som passar metoden. Figur 6.4 visar däremot att K-means kan ge dåliga resultat när klustren inte passar K-means antaganden, till exempel när klustren har en olämplig form eller storlek. Det är därför viktigt att komma ihåg att K-means inte automatiskt hittar den verkliga indelningen av data bara för att algoritmen hittar ett antal kluster.

## **Resonemangfrågor**

### **Hur kan man välja vilket antal kluster som ska användas för en K-means-modell?**

Det finns inget enda sätt som alltid ger rätt antal kluster, men man kan använda flera mått tillsammans. Ett vanligt sätt är att titta på `inertia`. Inertia beskriver hur nära observationerna ligger sina respektive centroid. En lägre inertia betyder alltså att observationerna ligger närmare sina centroid. Problemet är att inertia nästan alltid blir lägre när man ökar antalet kluster, så man kan inte bara välja det antal som har lägst inertia.
Därför kan man använda en så kallad elbow-metod. Man testar flera värden på `k` och ritar upp inertia mot antalet kluster. I början kan inertia minska mycket när man lägger till fler kluster. Sedan kommer en punkt där förbättringen blir betydligt mindre. Den punkten ser ut som en armbåge, eller ”elbow”, och kan vara ett rimligt val av antal kluster.

Ett annat mått är `silhouette score`. Det tar hänsyn både till hur nära en observation ligger observationerna i sitt eget kluster och hur långt bort den ligger från andra kluster. Ett högre silhouette score är bättre. Ett värde nära 1 betyder att observationen passar bra i sitt kluster och ligger långt från andra kluster, medan ett värde nära 0 betyder att den ligger nära gränsen mellan kluster. Negativa värden kan tyda på att observationen kanske passar bättre i ett annat kluster.

Man kan även använda ett `silhouette diagram`. Där visas silhouette-koefficienterna för observationerna i varje kluster. Varje kluster visas som en knivliknande form. Höjden på kniven visar hur många observationer klustret innehåller och bredden visar silhouette-koefficienterna för observationerna. Den streckade linjen visar det genomsnittliga silhouette-värdet. Diagrammet gör det därför möjligt att se både hur stora klustren är och hur väl observationerna passar in i sina kluster.
Därför är det bra att använda `inertia`, `silhouette score` och `silhouette diagram` tillsammans. De ger olika typer av information och kan tillsammans ge ett bättre underlag för att välja antal kluster.

### **Figur 6.10 på sidan 247 - Hur många kluster hade du valt och varför? Är det en ”exakt vetenskap” att välja antalet kluster?**

Jag hade valt ungefär fyra kluster utifrån figur 6.10. Inertia minskar kraftigt när antalet kluster ökar från 1 till 4. Efter `k = 4` fortsätter inertia att minska, men minskningen blir betydligt mindre. Det är därför ungefär vid `k = 4` som vi kan se en tydlig ”armbåge” i kurvan.
Det är däremot inte en exakt vetenskap att välja antalet kluster. Elbow-metoden ger ett underlag för beslutet, men det finns inte alltid ett enda självklart rätt svar. Man kan även använda andra mått, till exempel `silhouette score` och `silhouette diagram`, och ta hänsyn till vad klustringen ska användas till. Därför behöver man använda sitt eget omdöme när man väljer antal kluster.

### **Hur tolkar man figur 6.13 på sidan 251?**

Figur 6.13 visar silhouette diagrams för `k = 3`, `k = 4`, `k = 5` och `k = 6`. Den horisontella axeln visar silhouette coefficient och den röda streckade linjen visar det genomsnittliga `silhouette score`. Formen för varje kluster visar hur observationerna i klustret fördelar sig med avseende på sin silhouette-koefficient.
För `k = 3` och `k = 4` är klustren ganska ojämna i storlek. `k = 4` har ett relativt högt silhouette score, men klustren är inte helt jämnt fördelade. För` k = 5` blir klustren mer likvärdiga i storlek, även om det genomsnittliga silhouette score är något lägre än för `k = 4`.

För `k = 6 `blir resultatet sämre. Silhouette score är lägre och alla kluster ligger inte lika bra i förhållande till den streckade linjen. Det tyder på att vissa observationer inte passar lika bra i sina kluster.
I exemplet väljer boken därför `k = 5`, eftersom klustren då är mer likvärdiga i storlek. Samtidigt betonas att `k = 3` eller `k = 4` också hade kunnat vara rimliga val. Figuren visar alltså att valet av antal kluster inte bara handlar om att hitta det högsta silhouette score, utan även om att undersöka hur observationerna fördelas mellan klustren och vad klustringen ska användas till.
