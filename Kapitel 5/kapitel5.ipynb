# *Kapitel 5 - Dimensionsreducering*

## **Faktafrågor**

### **Vad menas med curse of dimensionality?**

Curse of dimensionality betyder att data blir svårare att arbeta med när antalet dimensioner, alltså antalet variabler, blir stort. När vi får fler dimensioner blir datautrymmet mycket större och det krävs därför mycket mer data för att täcka utrymmet på ett bra sätt. Observationerna kan bli mer utspridda, vilket gör det svårare för modellen att hitta tydliga mönster och generalisera till nya datapunkter.

Fler dimensioner kan också göra beräkningarna tyngre. I högdimensionell data kan avstånd mellan datapunkter dessutom bli mindre intuitiva. Det innebär att fler variabler inte automatiskt leder till bättre resultat. Om vissa variabler inte tillför tillräckligt mycket användbar information kan många dimensioner i stället göra problemet mer komplicerat.
Curse of dimensionality är därför en av anledningarna till att dimensionsreducering kan vara användbart. Genom att minska antalet dimensioner kan vi förenkla datan och samtidigt försöka behålla den viktigaste informationen.

### **Vad är dimensionsreducering och varför görs det?**

Dimensionsreducering betyder att man minskar antalet dimensioner eller variabler som används för att beskriva data, samtidigt som man försöker behålla så mycket viktig information som möjligt. Man kan alltså tänka att vi försöker förenkla datan utan att förlora det viktigaste mönstret.

Det görs bland annat för att data ska bli lättare att hantera och för att modeller kan behöva arbeta med färre dimensioner. Det kan minska mängden beräkningar och i vissa fall göra både träning och prediktion snabbare. Dimensionsreducering kan också göra data lättare att visualisera, till exempel genom att gå från många dimensioner till två dimensioner så att vi kan rita upp datan.

En annan anledning kan vara att minska mängden brus eller överflödiga variabler. Målet är alltså att få färre dimensioner samtidigt som så mycket viktig information som möjligt finns kvar. Dimensionsreducering kan dock också innebära att viss information går förlorad, så det är viktigt att väga fördelarna med färre dimensioner mot hur mycket information som försvinner.

### **Hur PCA fungerar. Använd figur 5.4 på sidan 224 i din förklaring.**

PCA står för Principal Component Analysis och är en metod för dimensionsreducering. Grundidén är att PCA letar efter nya riktningar, eller axlar, som beskriver variationen i datan så bra som möjligt. Dessa nya axlar kallas principalkomponenter.

I figur 5.4 ser vi ett dataset med två ursprungliga dimensioner, `X₁` och `X₂`. Punkterna ligger framför allt längs en diagonal riktning. PCA hittar då `C1`, som går i den riktning där variationen i datapunkterna är störst. `C1` blir därför den första och viktigaste principalkomponenten. `C2` är vinkelrät mot `C1` och beskriver variationen i den återstående riktningen.

Till höger i figuren visas hur datasetet i 2D projiceras på olika linjer i 1D. När datapunkterna projiceras på `C1` blir de mer utspridda längs den nya axeln `Z₁`. Det visar att `C1` fångar mycket av variationen i den ursprungliga datan. Projektionerna på de andra riktningarna ger mindre spridning. Figuren visar därför varför PCA väljer den riktning som fångar så mycket variation som möjligt.

PCA fortsätter sedan att identifiera ytterligare axlar tills antalet axlar motsvarar antalet ursprungliga variabler. De nya axlarna är ortogonala mot de tidigare och kallas principalkomponenter. Därefter kan man välja hur många principalkomponenter man vill behålla beroende på hur stor del av variansen man vill bevara.

I bokens exempel förklarar PK1 40 % av variansen, PK2 30 %, PK3 20 % och PK4 5 %. Tillsammans förklarar de 95 % av variansen, vilket innebär att man kan behålla PK1–PK4 om målet är att bevara minst 95 % av variansen. På så sätt kan man minska antalet dimensioner samtidigt som en stor del av variationen i datan behålls.

## **Resonemangfrågor**

### **Modellens prediktionsförmåga och effektivitet**

Kalle har en viktig poäng. Det är självklart viktigt att en modell gör bra prediktioner, men det är inte det enda som avgör vilken modell som passar bäst. I en verklig applikation måste man också tänka på hur lång tid modellen tar att träna och hur snabbt den kan göra en prediktion.

Tänk till exempel att två modeller nästan har samma träffsäkerhet, men att den ena tar mycket längre tid att träna och dessutom behöver betydligt längre tid för varje prediktion. Då kan den snabbare modellen vara ett bättre val, särskilt om modellen ska användas många gånger eller om användaren behöver få ett svar direkt. Tid kan alltså vara viktig både när modellen tränas och när den används.

Det kan även finnas andra praktiska saker att ta hänsyn till, som hur mycket minne och beräkningskraft modellen kräver och vad det kostar att köra den. Därför kan man inte alltid bara välja modellen med bäst prediktionsresultat. Man behöver väga prediktionskvaliteten mot tid och andra krav på den situation där modellen faktiskt ska användas.

### **Efter att vi genomfört en PCA, vad händer med tolkningen av variablerna?**

Efter PCA blir det svårare att tolka de ursprungliga variablerna direkt. Anledningen är att PCA skapar nya variabler, alltså principalkomponenter, i stället för att bara fortsätta använda originalvariablerna var för sig.

En principalkomponent är en kombination av flera av de ursprungliga variablerna. Det betyder att vi inte längre lika enkelt kan säga att en viss originalvariabel ensam har en viss betydelse för resultatet. I stället behöver vi titta på hur de ursprungliga variablerna bidrar till de olika principalkomponenterna.

PCA kan göra datan enklare att hantera och minska antalet dimensioner, men samtidigt blir resultatet mindre lätt att förklara för en människa. Vi får alltså en avvägning,vi kan vinna effektivitet och förenkla datan, men vi kan förlora en del av den direkta tolkningen av originalvariablerna.
