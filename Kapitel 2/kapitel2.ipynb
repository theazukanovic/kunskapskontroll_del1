# *Kapitel 2 - Ett ML projekt från början till slut*

## **Faktafrågor**

### **Beskriv de sju stegen i ett ML-projekt. Följs de i rak progression eller iterativt?**

Ett ML-projekt kan beskrivas med hjälp av en checklista med sju steg. Syftet med checklistan är att skapa struktur i arbetet, från att förstå problemet till att modellen används i verkligheten.

1. Definiera problemet och skapa en helhetsbild.
Först behöver man förstå vad man vill uppnå med projektet. Man behöver fundera på vad målet är, om lösningen kan bli användbar och hur man ska mäta om projektet lyckas. Det är också viktigt att fundera på vilka personer som berörs av projektet och vilka som behöver vara involverade.

2. Få tillgång till datan.
Därefter behöver man ta reda på vilken data som krävs och hur man får tillgång till den. Man behöver också hantera datan på ett lämpligt sätt, till exempel om den innehåller personuppgifter eller information som behöver anonymiseras. I detta steg är det också viktigt att tidigt lägga undan testdata, eftersom den ska användas först i slutet för att utvärdera modellens generaliseringsförmåga.

3. Utforska datan och göra en exploratory data analysis (EDA).
Här försöker man förstå datan bättre genom att använda exempelvis sammanfattande statistik och visualiseringar. Man kan undersöka om det finns mönster, felaktiga värden eller saknade värden i datan. EDA hjälper oss att upptäcka problem och få en bättre bild av vilken data vi faktiskt arbetar med innan vi börjar modellera.

4. Bearbeta datan.
I detta steg görs datan användbar för ML-modellering. Det kan innebära att hantera saknade värden, undersöka och eventuellt hantera outliers samt ta bort irrelevanta variabler. Här kan man också göra feature engineering, alltså skapa nya variabler eller förändra befintliga variabler för att de ska fungera bättre i modellen.

5. ML-modellering.
När datan är förberedd kan man börja skapa och träna modeller. I början är det rimligt att prova flera olika modeller med standardvärden för hyperparametrarna. Modellerna utvärderas sedan på valideringsdata för att se vilken modell som verkar fungera bäst. Man kan därefter justera hyperparametrar eller undersöka om andra variabler ger bättre resultat.

6. Presentera lösningen för intressenter.
När man har kommit fram till en möjlig lösning behöver resultatet presenteras för de personer som ska använda, förstå eller fatta beslut utifrån den. Presentationen behöver anpassas efter målgruppen. En tekniskt intresserad person kan behöva mer information om modellen, medan en chef eller annan beslutsfattare kanske främst behöver förstå resultatet, nyttan och eventuella konsekvenser.

7. Produktionssättning av modellen och övervakning av implementationen.
Om modellen ska användas i verkligheten behöver den produktionssättas. Det betyder att modellen går från utvecklingsmiljön till att börja användas i ett system, en applikation eller någon annan praktisk situation. Man behöver kontrollera att modellen fungerar som den ska och följa upp dess prestanda över tid. Data kan förändras, vilket kan göra att modellen fungerar sämre. Därför kan modellen behöva tränas om regelbundet med ny data.

Stegen följs vanligtvis inte i en helt rak progression. Ett ML-projekt är i stället iterativt, vilket innebär att man ofta går fram och tillbaka mellan stegen. Om man exempelvis upptäcker under arbetet att datan inte räcker för att nå det ursprungliga målet kan man behöva gå tillbaka, ändra målet eller försöka få tillgång till annan data. Checklistan är därför ett stöd i arbetet, inte en regel om att varje steg bara får göras en gång.

### **Vad menas med att en modell produktionssätts?**

Att produktionssätta en modell betyder att man tar den från utvecklings och testmiljön och gör den tillgänglig för att användas i en verklig situation. Modellen går alltså från att vara något man experimenterar med till att faktiskt användas av ett system, en applikation eller en användare.

Ett exempel är en modell som kan förutsäga priset på en bil. Under utvecklingen kan vi testa modellen i en notebook. När vi är nöjda kan modellen sparas och kopplas till en applikation där en användare matar in bilens egenskaper och får ett predikterat pris tillbaka.

Produktionssättning handlar därför inte bara om att spara en modellfil. Man behöver också se till att modellen kan köras på riktig data och att resten av systemet runt modellen fungerar.

Efter produktionssättningen behöver modellen följas upp. Om den data som modellen får i verkligheten förändras kan modellens resultat bli sämre. Därför är ett ML-projekt inte automatiskt färdigt bara för att modellen har satts i produktion.

### **Vad är scikit-learn för något?**

Scikit-learn är ett Python bibliotek för Machine Learning. Det innehåller färdiga verktyg för bland annat regression, klassificering, preprocessing, modellval, cross-validation och utvärdering. Biblioteket är uppbyggt på ett enhetligt sätt, vilket gör det enklare att arbeta med olika modeller och verktyg.

Scikit-learn följer fem centrala designprinciper. **Consistency** innebär att olika objekt i scikit-learn följer ett konsekvent och enhetligt API. Många objekt använder exempelvis metoder som `fit()`, `predict()` och `transform()`. **Inspection** innebär att parametrar och värden som modellen lär sig under träningen görs tillgängliga som attribut. Det gör det möjligt att undersöka vad modellen har lärt sig. **Non-proliferation of classes** innebär att scikit-learn försöker undvika att skapa många olika typer av specialklasser. Datan representeras i stället med exempelvis NumPy-arrayer, vilket gör biblioteket enklare att kombinera med andra Python-bibliotek. **Composition** innebär att olika delar av ett Machine Learning-flöde kan kombineras och byggas ihop av befintliga komponenter. Exempelvis kan transformers och modeller kombineras i ett gemensamt flöde. **Sensible defaults** innebär att många parametrar har rimliga standardvärden, vilket gör att modeller ofta kan användas utan att alla parametrar behöver anges manuellt.

**Estimators** är objekt som kan lära sig något från data genom metoden `fit()`. Ett exempel är `LinearRegression`, där `fit()` används för att anpassa modellen till träningsdatan. **Predictors** är estimators som kan använda det de har lärt sig för att göra prediktioner. De använder bland annat `predict()` för att förutsäga värden eller klasser för nya observationer. **Transformers** används för att omvandla data. De använder ofta `fit()` för att lära sig hur transformeringen ska göras och `transform()` för att genomföra den. Ett exempel är att standardisera numeriska variabler eller omvandla data inför modellträning.

De tre gränssnitten hänger alltså ihop. En estimator lär sig från data, en predictor kan använda det den lärt sig för att göra prediktioner och en transformer omvandlar data inför nästa steg.

### **Vad är TensorFlow och Keras?**

TensorFlow är ett ramverk som används för maskininlärning och framför allt neurala nätverk och Deep Learning. Det kan hantera de beräkningar som behövs när modeller tränas på data.

Keras är ett högnivå-API som gör det enklare att bygga och träna neurala nätverk. Keras kan användas tillsammans med TensorFlow och ger ett mer lättillgängligt sätt att beskriva modellens lager och träning.

Man kan förenklat tänka att TensorFlow är den större tekniska plattformen för beräkningarna, medan Keras ger ett enklare gränssnitt för att bygga neurala nätverk.

Det är relevant att känna till skillnaden eftersom scikit-learn används mycket för klassiska ML-modeller och verktyg, medan TensorFlow och Keras är särskilt förknippade med neurala nätverk och Deep Learning.

### **Vad säger du om deras dialog?**

Stina har rätt i sak. Problemet med Kalles sätt är att han använder testdatan för att fatta beslut om hur modellen ska ändras. Testdatan ska egentligen sparas till slutet och användas som en så objektiv slutkontroll som möjligt.

Tanken med testdata är att modellen inte ska ha anpassats efter den. Om Kalle testar en modell på testdatan, ser resultatet, ändrar modellen och sedan testar igen, använder han indirekt testdatan för att styra modellens utveckling.

Om han upprepar detta många gånger kan modellen till slut bli anpassad till just testdatan. Då är testdatan inte längre ett neutralt test av hur modellen fungerar på ny data. Testresultatet kan därför se bättre ut än modellens verkliga förmåga att generalisera.

Kalle har alltså rätt i att man ska förbättra modellen om den inte är tillräckligt bra, men han använder fel data för att göra det. Han bör i stället använda träningsdata för träningen och valideringsdata eller cross-validation för att jämföra modeller och justera hyperparametrar.

När han sedan är klar med sina val kan testdatan användas på slutet för att få en så rättvis uppskattning som möjligt av hur modellen fungerar på data som den inte har sett.

Det viktiga är därför att testdatan hålls så oberoende som möjligt. Annars försvinner en stor del av poängen med att ha en separat testuppsättning.

### **Varför misslyckas många AI/ML-projekt och hur bör vi förhålla oss till det?**

Många AI- och ML-projekt når inte sina ursprungliga mål, och det kan bero på flera olika saker. Ett problem kan vara att det finns för lite data för att modellen ska kunna tränas tillräckligt bra.
Datan kan också vara icke-representativ. Det betyder att träningsdatan inte liknar den data som modellen senare ska användas på. Då kan modellen fungera sämre när den möter ny data.

Dålig datakvalitet är också ett vanligt problem. Datan kan innehålla fel, vara slarvigt insamlad eller ha saknade värden. Då blir det svårt för modellen att lära sig de mönster som egentligen finns.

Ett annat problem är irrelevanta features. Om de variabler som används inte är relevanta för problemet blir det svårt att skapa en bra modell. Det finns ett uttryck som säger “shit in-shit out”, vilket betyder att dålig eller irrelevant data ofta leder till dåliga resultat.

Det finns också en risk för överanpassning eller underanpassning. En överanpassad modell fungerar bra på träningsdatan men sämre på ny data. En underanpassad modell är däremot för enkel för att fånga viktiga mönster.
Därför bör vi ha rimliga förväntningar och arbeta iterativt. Ett projekt behöver inte vara helt misslyckat bara för att det inte uppnår sitt första mål. Man kan fortfarande få fram något användbart, till exempel kunskap om datan, problemet eller vilka förändringar som behövs i nästa steg.
