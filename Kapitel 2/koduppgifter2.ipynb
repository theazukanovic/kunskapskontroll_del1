## **8. Förklara vad koden nedan gör. Varför är det viktigt att kunna spara en modell?**

Koden börjar med att importera make_regression, LinearRegression samt dump och load från joblib. make_regression används för att skapa ett artificiellt regressionsdataset som vi kan träna en modell på.

När `X, y = make_regression(n_samples=20000, n_features=3, noise=0.1)` körs skapas data med 20 000 observationer och tre features. X innehåller indata och y innehåller värdena som modellen ska lära sig att förutsäga.

Sedan skapas och tränas en linjär regressionsmodell med `LinearRegression().fit(X, y)`. Efter fit() har modellen lärt sig sina parametrar från datan.

Med `dump(model, "linear_model.joblib")` sparas den tränade modellen i en fil. Vi behöver då inte träna modellen från början varje gång vi vill använda den.

Med `model_loaded = load("linear_model.joblib")` läser vi tillbaka modellen från filen. Den laddade modellen kan sedan användas igen.

Till sist används `model_loaded.predict(X[:5])` för att göra prediktioner på de fem första observationerna. Poängen är att prediktionerna görs med den sparade och laddade modellen.

Att kunna spara en modell är viktigt i ett riktigt ML-projekt eftersom träning kan ta tid och modellen ofta behöver användas senare i exempelvis en applikation eller ett API. Man tränar därför modellen, sparar den och laddar den när den behövs.

Det är också en viktig del av att produktionssätta en modell, eftersom den tränade modellen behöver kunna användas utanför själva utvecklingsnotebooken.

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from joblib import dump, load
```

```python
X, y = make_regression(n_samples=20000, n_features=3, noise=0.1)
model = LinearRegression().fit(X, y)
dump(model, "linear_model.joblib")
model_loaded = load("linear_model.joblib")
print(model_loaded.predict(X[:5]))
```

## **9. data_01.csv**

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error
```

### **Load data**

```python
# a) Läs in datasetet
df = pd.read_csv("dataset/data_01.csv")
```

```python
df.head()
```

#### **Define Features and Target**

```python
# b) Dela upp datasetet i X och y

X = df.drop(columns="target")
y = df["target"]
```

```python
X.head()
```

```python
y.head()
```

#### **Train, Validation and Test Split**

```python
# c) Dela upp datasetet i tränings, validerings och testdata

X_train_full, X_test, y_train_full, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_train_full, y_train_full, test_size=0.15, random_state=42)

print("Träningsdata:", X_train.shape)
print("Valideringsdata:", X_val.shape)
print("Testdata:", X_test.shape)
```

#### **Train Regression Models**

```python
# d) Träna två regressionsmodeller

linreg = LinearRegression()
tree = DecisionTreeRegressor(random_state=42)

linreg.fit(X_train, y_train)
tree.fit(X_train, y_train)
```

#### **Evaluate Models**

```python
# e) Utvärdera modellerna på valideringsdatan

linreg_pred = linreg.predict(X_val)
tree_pred = tree.predict(X_val)

linreg_rmse = root_mean_squared_error(
    y_val,
    linreg_pred
)

tree_rmse = root_mean_squared_error(
    y_val,
    tree_pred
)

print("RMSE Linear Regression:", linreg_rmse)
print("RMSE Decision Tree:", tree_rmse)
```

#### **Select the Best Model**

```python
# f) + g) Välj den bästa modellen och träna om på träning och valideringsdatan

if linreg_rmse < tree_rmse:
    best_model = LinearRegression()
    best_model_name = "Linear Regression"
else:
    best_model = DecisionTreeRegressor(random_state=42)
    best_model_name = "Decision Tree"

print("Best model:", best_model_name)
```

#### **Retrain on Training and Validation Data**

```python
best_model.fit(X_train_full, y_train_full)
```

#### **Final Test Evaluation**

```python
test_pred = best_model.predict(X_test)

test_rmse = root_mean_squared_error(
    y_test,
    test_pred
)

print("Final Test RMSE:", test_rmse)
```

#### **Train Final Model on the Full Dataset**

```python
# h) Träna modellen på hela datasetet
best_model.fit(X, y)
```

## **10. Datasetet salary_dataset.csv**

```python
import pandas as pd

from sklearn.model_selection import train_test_split, cross_validate
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error
```

### **Load data**

```python
# a) Läs in datasetet

df = pd.read_csv("dataset/salary_dataset.csv")
```

### **Define features and target**

```python
# a) Läs in datasetet och dela upp i X och y

X = df[["YearsExperience"]]
y = df["Salary"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

### **Train-Test Split**

```python
# b) Dela upp datan i tränings- och testdata

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Test data:", X_test.shape)
```

### **Cross-Validation**

```python
# c) Träna två regressionsmodeller med 5-delad korsvalidering

linreg = LinearRegression()

tree = DecisionTreeRegressor(
    max_depth=3,
    random_state=42
)

linreg_cv = cross_validate(
    linreg,
    X_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)

tree_cv = cross_validate(
    tree,
    X_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)
```

### **Compare Models**

```python
linreg_rmse = -linreg_cv["test_score"].mean()
tree_rmse = -tree_cv["test_score"].mean()

print("Linear Regression CV RMSE:", linreg_rmse)
print("Decision Tree CV RMSE:", tree_rmse)
```

### **Select the Best Model**

```python
if linreg_rmse < tree_rmse:
    best_model = linreg
    best_model_name = "Linear Regression"
else:
    best_model = tree
    best_model_name = "Decision Tree"

print("Best model:", best_model_name)
```

### **Final Test Evaluation**

```python
# d) Välj den bästa modellen och utvärdera den på testdata
best_model.fit(X_train, y_train)

pred = best_model.predict(X_test)

test_rmse = root_mean_squared_error(
    y_test,
    pred
)

print("Best model:", best_model_name)
print("Test RMSE:", test_rmse)
```

### **Analysis**

Linear Regression fick ett lägre RMSE än Decision Tree vid den 5-delade korsvalideringen. Därför valdes Linear Regression som den bäst presterande modellen.

Vid den slutliga utvärderingen på testdatan fick modellen ett RMSE på cirka 7059. Det innebär att modellens prediktioner i genomsnitt har ett fel på ungefär 7059 löneenheter på testdatan.

Resultaten visar att sambandet mellan `YearsExperience` och `Salary` i detta dataset kan modelleras relativt väl med en linjär modell.

## **11. Datasetet mpg**

```python
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
```

### **Load data**

```python
# a) Läs in datasetet

df = sns.load_dataset("mpg")
```

```python
df.head()
```

### **Handle missing values**

```python
# b) Droppa rader med saknade värden

df = df.dropna()
```

### **Remove the name column**

```python
# c) Droppa kolumnen name

df = df.drop(columns="name")
```

### **Encode categorical data**

```python
# d) Utför dummy-variable-encoding på origin

df = pd.get_dummies(df, columns=["origin"], drop_first=True)
```

```python
df.head()
```

### **Define features and target**

```python
# e) Dela upp datasetet i X och y

X = df.drop(columns="mpg")
y = df["mpg"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

### **Train-Test Split**

```python
# f) Dela upp datasetet i tränings och testdata
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training data:", X_train.shape)
print("Test data:", X_test.shape)
```

### **Train linear regression**

```python
# g) Träna en linjär regressionsmodell på träningsdatan

model = LinearRegression()
model.fit(X_train, y_train)
```

### **Evaluate the model**

```python
pred = model.predict(X_test)

rmse = root_mean_squared_error(
    y_test,
    pred
)

print("Test RMSE:", rmse)
```

### Analysis

Modellen tränades med de numeriska variablerna samt de två dummy-variablerna för `origin`. På testdatan fick modellen ett RMSE på cirka 3,26.

Det innebär att modellens prediktioner i genomsnitt har ett fel på ungefär 3,26 mpg. Resultatet visar att en linjär regressionsmodell kan användas för att förutsäga bränsleeffektiviteten utifrån de variabler som finns i datasetet.

## **12 – Förbättra Random Forest Regression på huspriser**

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error
```

### **Load data**

```python
df = pd.read_csv("dataset/housing.csv")
```

```python
df.head()
```

### **Prepare data**

```python
housing = pd.get_dummies(df, columns=["ocean_proximity"], drop_first=True)
```

```python
X = housing.drop(columns=["median_house_value"])
y = housing["median_house_value"]
```

### **Train, Validation and Test Split**

```python
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.2, random_state=40
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.25, random_state=36
)
```

### **Handle Missing Values**

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_val_imputed = imputer.transform(X_val)
```

### **Baseline Model**

```python
baseline_model = RandomForestRegressor(random_state=42)

baseline_model.fit(X_train_imputed, y_train)
baseline_pred = baseline_model.predict(X_val_imputed)
baseline_rmse = root_mean_squared_error(y_val, baseline_pred)

print("Baseline Random Forest RMSE:", baseline_rmse)
```

### **Feature Selection**

```python
X_train_selected = X_train.drop(columns=["housing_median_age"])
X_val_selected = X_val.drop(columns=["housing_median_age"])
X_test_selected = X_test.drop(columns=["housing_median_age"])
```

```python
selected_imputer = SimpleImputer(strategy="median")

X_train_selected = selected_imputer.fit_transform(X_train_selected)
X_val_selected = selected_imputer.transform(X_val_selected)
X_test_selected = selected_imputer.transform(X_test_selected)
```

#### **Random Forest with Feature Selection**

```python
selected_model = RandomForestRegressor(n_estimators=100, random_state=42)

selected_model.fit(X_train_selected, y_train)
selected_pred = selected_model.predict(X_val_selected)
selected_rmse = root_mean_squared_error(y_val, selected_pred)

print("Random Forest with Feature Selection RMSE:", selected_rmse)
```

### **Hyperparameter Experiment**

```python
tuned_model = RandomForestRegressor(n_estimators=300, random_state=42)

tuned_model.fit(X_train_selected, y_train)
tuned_pred = tuned_model.predict(X_val_selected)
tuned_rmse = root_mean_squared_error(y_val, tuned_pred)

print("Random Forest with Feature Selection and 300 Trees RMSE:", tuned_rmse)
```

### **Model Comparison**

```python
results = pd.DataFrame(
    {
        "Model": [
            "Baseline Random Forest",
            "Random Forest + Feature Selection",
            "Random Forest + Feature Selection + 300 Trees",
        ],
        "RMSE": [baseline_rmse, selected_rmse, tuned_rmse],
    }
)

results
```

### **Final Model Evaluation**

```python
final_model = RandomForestRegressor(n_estimators=300, random_state=42)

X_train_full_selected = X_train_full.drop(columns=["housing_median_age"])
X_test_selected = X_test.drop(columns=["housing_median_age"])

final_imputer = SimpleImputer(strategy="median")

X_train_full_selected = final_imputer.fit_transform(X_train_full_selected)
X_test_selected = final_imputer.transform(X_test_selected)

final_model.fit(X_train_full_selected, y_train_full)
final_pred = final_model.predict(X_test_selected)
final_rmse = root_mean_squared_error(y_test, final_pred)

print("Final Model RMSE on Test Data:", final_rmse)
```

### **Analysis**

Jag började med en Random Forest-modell som baseline och använde samma typ av train-, validation- och testuppdelning som i bokens kodexempel. Baseline-modellen gav ett RMSE på cirka 49 768 på valideringsdatan.

Därefter testade jag variabelselektion genom att ta bort `housing_median_age`. Detta gav ett RMSE på cirka 48 868, vilket var en förbättring jämfört med baseline-modellen.

Jag testade sedan att öka antalet träd i Random Forest från standardinställningen till 300 träd. Tillsammans med variabelselektionen gav detta ett RMSE på cirka 48 802 på valideringsdatan.

Resultaten visar att både variabelselektionen och ökningen av antalet träd förbättrade resultatet på valideringsdatan i dessa experiment. Förbättringen från 100 till 300 träd var däremot relativt liten, vilket visar att fler träd inte automatiskt ger en stor förbättring.

Den slutliga modellen tränades därefter på träningsdatan och utvärderades på den separata testdatan. Modellen fick ett RMSE på cirka 50 048 på testdatan.
